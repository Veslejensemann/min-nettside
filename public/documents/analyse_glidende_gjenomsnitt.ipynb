```python
import yfinance as yf
import pandas as pd

tickere_OSEBX = ["EQNR.OL", "YAR.OL", "NHY.OL", "ELK.OL", "MOWI.OL", "DNB.OL", "AKERBP.OL", "KOG.OL", "NODL.OL", "REC.OL", "TOM.OL", "SALM.OL", "STB.OL", "SUBC.OL", "TEL.OL"]   #velg tickere du vil se på Oslo Børs
data = yf.download(tickere_OSEBX, period="5y")["Close"] #bruk funbksjonen til yfinance, og hent lengden på perioden til dataene
print(data["EQNR.OL"])
```

```python
resultater = {} #opprett en tom dictionary for å lagre resultatene

for ticker in tickere_OSEBX:
    if ticker not in data.columns:
        continue
    df = data[[ticker]].copy()
    df.columns = ["Kurs"]
    df["GJ50"] = df["Kurs"].rolling(window=50).mean() # Beregn glidende gjennomsnitt for 50 dager
    df["GJ200"] = df["Kurs"].rolling(window=200).mean() # Beregn glidende gjennomsnitt for 200 dager

    resultater[ticker] = df 
```

```python
print(resultater["YAR.OL"]) 
```

```python
import plotly.express as px

fig = px.line(
    resultater["YAR.OL"], #NB: velg ticker for selskapet du vil se på
    y=["Kurs", "GJ50", "GJ200"],
    title="Kurs og glidende gjennomsnitt"
) #opprett graf
fig.show()
```

```python
with pd.ExcelWriter("gg_oslo_bors_analyse.xlsx", engine="openpyxl") as writer:
    for ticker, df in resultater.items():
        df.to_excel(writer, sheet_name=ticker)

#Skriv ut i excel om ønskelig        
```
