```python
# Triangulær arbitrasje

#Banken gi disse prisene
"EUR/USD = 0.92 EUR"
"NOK/EUR = 11.65 NOK"
"USD/NOK = 0.098 USD"

#Du har så mye å handle for
"Startbeøp = 1000 USD"
```

```python
#Sjekker markedet
"EUR/USD = EUR/NOK * NOK/USD = 1/11.65 * 1/0.098 = 0.875 EUR -> Underprisent i banken"

#Kjøp EURO
"1000 USD * 0.92 EUR/USD = 920 EUR"

#Konverterer til NOK
"920 EUR * 11.65 NOK/EUR = 10718 NOK"

#Konverter tilbake til USD
"10718 NOK * 0.098 USD/NOK = 1050 USD"

#profit
"1050 USD - 1000 USD = 50 USD" # -> 5% profit
```

```python
#------------- KODE -----------------
euro_per_usd = 0.92
usd_per_euro = 1 / euro_per_usd

nok_per_euro = 11.65
euro_per_nok = 1 / nok_per_euro

usd_per_nok = 0.098
nok_per_usd = 1 / usd_per_nok

start_beløp_usd = 1000

# --- Det finnes bare 2 mulige sirkler ---

# Rute 1: USD -> EUR -> NOK -> USD
produkt_1 = euro_per_usd * nok_per_euro * usd_per_nok

# Rute 2 (motsatt vei): USD -> NOK -> EUR -> USD
produkt_2 = nok_per_usd * euro_per_nok * usd_per_euro

print(f"Rute 1 (USD->EUR->NOK->USD): produkt = {produkt_1:.4f}")
print(f"Rute 2 (USD->NOK->EUR->USD): produkt = {produkt_2:.4f}")

# --- Velg den beste ---
if produkt_1 > produkt_2:
    beste_produkt = produkt_1
    beste_rute = "USD -> EUR -> NOK -> USD"
else:
    beste_produkt = produkt_2
    beste_rute = "USD -> NOK -> EUR -> USD"

sluttbeløp = start_beløp_usd * beste_produkt
profitt_prosent = (beste_produkt - 1) * 100

print()
print(f"Beste rute: {beste_rute}")
print(f"Sluttbeløp: {sluttbeløp:.2f} USD")
print(f"Profitt: {profitt_prosent:.2f} %")
```
